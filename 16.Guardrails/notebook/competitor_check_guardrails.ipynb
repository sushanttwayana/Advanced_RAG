```python
from guardrails import Guard
from guardrails_ai.competitor_check import CompetitorCheck

validator = CompetitorCheck(
    competitors=["Apple", "Samsung"],
    on_fail="exception"
)

guard = Guard().use(validator)

response = guard.validate(
    "I like all the fruits, but apple and grapes are my favorite."
)

print(response)
```

```python
from guardrails import Guard
from guardrails_ai.competitor_check import CompetitorCheck

validator = CompetitorCheck(
    competitors= ["Apple", "Google", "Microsoft", "Samsung", "Amazon"],
    on_fail="exception"
)

print("Validator:", validator)
print("on_fail:", validator.on_fail_descriptor)

guard = Guard().use(validator)

print("Guard created")

user_input_1 = "I like all the fruits, but apple and grapes are my favorite."
user_input_2 = "Apple is great, makes durable and high quality products."

try:
    response = guard.validate(user_input_2)

    print("\nResponse:")
    print(response)

    print("\nValidation passed:")
    print(response.validation_passed)

    print("\nValidated output:")
    print(response.validated_output)

except Exception as e:
    print("\nEXCEPTION:")
    print(type(e).__name__)
    print(e)
```
